## What is a Series?
A Pands Series is like a column in table which a one-dimensional array holding data of any type.

```python
a = [1, 7, 2, 5]
myVar = pd.Series(a)
print(myVar)
```

## Labels
If nothing else is specified the value are labeled with their index number.

```python
print(myVar[0])
```

### Create Label

```python
b = [1, 7, 2, 5]    
myVar_v2 = pd.Series(b, index=['x','y','z','w'])
print(myVar_v2)
```

```python
# access an item by referring to the label.
print(myVar_v2['z'])
```

### Key/Value Objects as Series
Also a key/value object like dictionary when creating a series

```python
calories = {'day_1':420, 'day_2':380,'day_3':390, 'day_4':410}
myVar_v3 = pd.Series(calories)
print(myVar_v3)
```

#### Create a Series using only data from "day1" and "day2":

```python
myVar_day_1_2 = pd.Series(calories, index=['day_1', 'day_2'])
print(myVar_day_1_2)
```

## DataFrames

```python
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}

myVar_df = pd.DataFrame(data)
print(myVar_df)
```
