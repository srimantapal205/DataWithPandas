#  Pandas - Cleaning Data

Data cleaning means fixing the data in the datset. Like as

* Empty cells
* Data in wrong format
* Wrong Data
* Duplicates


```python
import pandas as pd

df = pd.read_json('../data/data_v2.json')

print(df.to_string())
```

* In this data set contains some empty data
* Data set contain wrong data format
* data set contain wron data
* remove duplicate data

### Empty Cells

Empty cell can give wrong result.


```python
# Removes empty cell and create new data set
new_df = df.dropna()
print(new_df.to_string())
```

### Replace Empty Values

```python
#Replace NULL values with the number 130:
replace_df = df.fillna(130)
print(replace_df.to_string())
```
